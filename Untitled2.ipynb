```python
import pandas as pd
```

```python
sample_data = {
"Name": ["Salma", "Babu", "Nand", "Irfan"],
"Math": [78, 34, 29, 33],
"English":[58, 74, 69, 93],
"Science":[68, 54, 59, 91],
"History":[78, 64, 50, 41],
}
Students = pd.DataFrame(sample_data)
Students.to_csv("students.csv", index = False)
new_data = pd.read_csv("students.csv")
print(new_data)
```

```python
print(Students["Math"]) #selecting one column
print(Students[["Name", "Math"]]) #selecting multiple columns
math_scores = Students["Math"]
print(math_scores)
```

Selecting Rows: (loc[] selects rows using labels

```python
print(Students.loc[0]) #Selecting one row
print(Students.loc[[0, 1,]]) #Selecting multiple rows
```

```python
print(Students.loc[0]) #Selecting one row
print(Students.loc[[0, 1,]]) #Selecting multiple row
#iloc[] selects rows using position numbers
print(Students.iloc[0]) #Selecting one row
print(Students.iloc[[0, 1,]]) #Selecting multiple rows
```

```python
#selecting specific rows and columns
print(Students.loc[0, "Name"]) #Selecting one rows and one column
print(Students.loc[[0,1], ["Name", "Math"]]) #multiple rows and multiple columns
```

```python
#Filtering data
#students with more than 60 in maths
highscores = Students[Students["Math"]>60]
print(highscores)
print(Students["Math"].mean()) #average score in maths
print(Students["Math"].min()) #minimum score in maths
print(Students["Math"].max()) #maximum score in maths
print(Students["Math"].sum()) #total score in maths

```

create a dataframe called spotify with the following columns:
track_name, artist, genre, duration_mins, and popularity_score. enter hypothetical data. then display the first row, find the most popular track, find the verage duration, and display tracks with a popularity of more than 70/100

```python
import pandas as pd
data = {
    "track_name":["Blinding lights", "Levitating", "Perfect", "Flowers"],
    "artist":["The Weekend", "Dua Lipa", "Ed Sheeran", "Miley Cyrus"],
    "genre":["pop", "pop", "pop", "pop"],
    "duration mins":[3.2, 3.4, 4.2, 3.3],
    "popularity_score": [95, 80, 90, 75]
}
spotify = pd.DataFrame(data)
print(spotify.head(1))
print(spotify[spotify["popularity_score"]==spotify["popularity_score"].max()])
print(spotify["duration mins"].mean())
print(spotify[spotify["popularity_score"]>70])
```

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
students = pd.DataFrame({
    "students":["John", "Mary", "Ali", "Grave",
                "James", "Faith", "Brian", "Lucy"],
    "Age":[20,21,22,20,23,21,22,20],
    "Gender":["Male", "Female", "Male", "Male",
              "Male", "Female", "Male", "Female"],
    "Math":[78,85,90,66,74,88,82,91],
    "Science":[80,79,95,70,88,90,76,90],
    "English":[75,88,92,72,81,86,79,90]
})
```

```python
print(students)
```

```python
#A line chart shows trends over time or ordered data.
plt.plot(students["students"], students["Math"])
plt.show()
```

```python
#Adding titles
plt.plot(students["students"], students["Math"])
plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

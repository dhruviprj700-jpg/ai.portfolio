#**Data Frames**
A data frame is a two dimensial table. it contains rows and columns. similar to an excel spreadsheet

```python
data = {
    "Name": ["Salma", "Babu", "Nand", "Irfan"],
    "Math": [78, 34, 29, 33],
    "English": [58, 74, 69, 93],
    "Science": [68, 54, 59, 91]
    }
students = pd.DataFrame(data)
print(students)
```

```python
students = pd.DataFrame(students) # Convert the dictionary to a DataFrame
print(students.dtypes)# Checking data types
print(students.shape)# Checking the shape of the data frame
```

Array indexing

```python
students = np.array(["Irena", "Irfan", "Moksh", "Rudra", "Stanley", "Levi"])
print(students[0])
print(students[-1])
```

Array slicing:slicing means choosing multiple elements

```python
print(students[1:4])
print(students[:3])# Print first 3 elemets
print(students[-2])# print last 2 elemets
print(students[::2])# skip values
```

```python
marks = np.array([14, 34, 56, 89, 45, 35, 67, 56, 54, 43])
print(marks[0])
print(marks[-1])
print(marks[2:5])
print(marks[::2])
```

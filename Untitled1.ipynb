Array funvtions

```python
import numpy as np
marks = np.array([34, 27, 35, 20, 10])
print(np.min(marks))#minimum mark
print(np.max(marks))#maximum mark
print(np.mean(marks))#average mark
print(np.sum(marks))#total marks
print(np.sort(marks))#sort
print(np.median(marks))#median
print(np.std(marks))#standard deviation
```

Reshaping Arrays

```python
numbers = np.array([1, 2, 3, 4, 5, 6, 7, 8])
reshape_numbers = numbers.reshape(2, 4)#reshape(2 rows, 4 columns)
print(reshape_numbers)
print(reshape_numbers[0,1])#Accessing elements
print(reshape_numbers[1])#Accessing a row
print(reshape_numbers[:, 1])#Accessing a column
```

```python
import numpy as np
scores = np.random.randint(0, 101, 15)
print("student scores:", scores)
print("average:", np.mean(scores))
print("highest mark:", np.max(scores))
print("lowest mark", np.min(scores))
print("sorted marks:", np.sort(scores))
```
